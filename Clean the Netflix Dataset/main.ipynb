```python
###Aufgabe 2 - Identify and handle missing values column by column###

import pandas as pd
df = pd.read_csv("netflix_titles.csv")

# Identifizieren
df.isnull().sum()
df.isnull().sum() / len(df)

# Mit anderen Werten ersetzen
df.fillna("<Unbekannter Wert>", inplace=True)

```

```python
###Aufgabe 3 - Fix mixed-type columns (e.g., duration stored as "90 min")###

import pandas as pd

df = pd.read_csv('netflix_titles.csv')

# Zahl aus dem String extrahieren
df['duration_int'] = df['duration'].str.extract(r'(\d+)').astype('Int64')

# Je nach Typ die richtige Spalte befüllen
df['duration_minutes'] = df['duration_int'].where(df['type'] == 'Movie')
df['duration_seasons'] = df['duration_int'].where(df['type'] == 'TV Show')

# Hilfsspalte aufräumen
df.drop(columns=['duration_int'], inplace=True)

print(df[['type', 'duration', 'duration_minutes', 'duration_seasons']])
```

```python
###Aufgabe 4 - Parse date columns into proper datetime objects###

import pandas as pd

df = pd.read_csv("netflix_titles.csv")

print(df['date_added'].head())

# Konvertierung zu "Jahr - Monat(int) - Tag"
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), format='%B %d, %Y')

print(df['date_added'])
```

```python
###Aufgabe 5 - Export the cleaned DataFrame to a new CSV file###

#### Aufgabe 1 - Inspect the DataFrame with .info(), .describe(), and .head()####
print("=== .info() ===")
df.info()

print("\n=== .describe() ===")
print(df.describe())

print("\n=== .head() ===")
print(df.head())


#### Aufgabe 2 - Identify and handle missing values column by column####
print("\n=== Fehlende Werte vor der Bereinigung ===")
print(df.isnull().sum())

print("\n=== Anteil fehlender Werte ===")
print(df.isnull().sum() / len(df))

df["director"] = df["director"].fillna("Unbekannt")
df["cast"] = df["cast"].fillna("Unbekannt")
df["country"] = df["country"].fillna("Unbekannt")


#### Aufgabe 3 - Fix mixed-type columns (e.g., duration stored as "90 min")####

df["duration_int"] = (
    df["duration"]
    .str.extract(r"(\d+)", expand=False)
    .astype("Int64")
)

df["duration_minutes"] = df["duration_int"].where(
    df["type"] == "Movie"
)

df["duration_seasons"] = df["duration_int"].where(
    df["type"] == "TV Show"
)

df.drop(columns=["duration_int"], inplace=True)

print("\n=== Dauer-Spalten ===")
print(
    df[
        [
            "type",
            "duration",
            "duration_minutes",
            "duration_seasons"
        ]
    ].head(10)
)


####Aufgabe 4 - Parse date columns into proper datetime objects####

print("\n=== Datum vor der Konvertierung ===")
print(df["date_added"].head())

df["date_added"] = pd.to_datetime(
    df["date_added"].str.strip(),
    format="%B %d, %Y",
    errors="coerce"
)

print("\n===Datum nach der Konvertierung===")
print(df["date_added"].head())


####Aufgabe 5 - Export the cleaned DataFrame to a new CSV file####

df.to_csv(
    "netflix_titles_cleaned.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nFertig!")
print("Datei gespeichert als: netflix_titles_cleaned.csv")
print(f"Zeilen: {len(df)} | Spalten: {len(df.columns)}")
```
